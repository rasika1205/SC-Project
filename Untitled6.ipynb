```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.preprocessing import StandardScaler
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers

print("TensorFlow version:", tf.__version__)
```

```python
SEED = 42

np.random.seed(SEED)
tf.random.set_seed(SEED)
```

```python
DATA_PATH = "sfpark_feature_engineered_dataSet.csv"

df = pd.read_csv(DATA_PATH)

df["START_TIME_DT"] = pd.to_datetime(df["START_TIME_DT"])

print("Shape:", df.shape)
print(df.head())
```

feature list

```python
FEATURES = [
    "GMP_OCCUPANCY",
    "GMP_AVAILABILITY",
    "OP_OCCUPANCY",
    "OP_AVAILABILITY",
    "TOTAL_OCCUPANCY",
    "TOTAL_UNKNOWN_RATE",
    "OP_UNKNOWN_RATE",
    "GMP_UNKNOWN_RATE",
    "EFFECTIVE_MONITORED_SPACES",
    "NONOP_RATE",
    "GMP_COMM_SHARE",
    "RATE",
    "RATE_CHANGE",
    "HOUR_SIN",
    "HOUR_COS",
    "DOW_SIN",
    "DOW_COS",
    "WEEKEND",
    "GMP_OCC_LAG_1",
    "GMP_OCC_LAG_2",
    "GMP_OCC_LAG_3",
    "GMP_OCC_CHANGE_1",
    "GMP_OCC_CHANGE_3",
    "GMP_ROLLING_MEAN_3"
]

TARGET = "TARGET_GMP_AVAILABILITY_1H"

LOOKBACK = 3
```

We verify that everything exists

```python
required_columns = (
    ["BLOCK_ID", "START_TIME_DT"]
    + FEATURES
    + [TARGET]
)

missing = [
    col for col in required_columns
    if col not in df.columns
]

if missing:
    raise ValueError(
        "Missing columns:\n" + "\n".join(missing)
    )

print("All required columns exist.")
```

filter model-ready data

```python
if "MODEL_INPUT_COMPLETE" in df.columns:
    model_ready = df[
        df["MODEL_INPUT_COMPLETE"]
        & df["HISTORY_CONTIGUOUS"]
        & df["TARGET_CONTIGUOUS"]
        & df["TARGET_GMP_QUALITY_OK"]
        & df["QUALITY_OK"]
        & df["GMP_QUALITY_OK"]
    ].copy()
else:
    model_ready = df.copy()

print("Model-ready shape:", model_ready.shape)
```

Sort the data

```python
model_ready = model_ready.sort_values(
    ["BLOCK_ID", "START_TIME_DT"]
).reset_index(drop=True)
```

The train test split is done in this manner instead of using train_test_split since we are dealing with time-series data.


Past Data → Training Set

More Recent Data → Validation Set

Latest Data → Test Set

```python
min_date = model_ready["START_TIME_DT"].min()
max_date = model_ready["START_TIME_DT"].max()

print("Data:", min_date, "→", max_date)
```

```python
train_end = pd.Timestamp("2012-06-30")
val_end = pd.Timestamp("2012-12-31")

train_df = model_ready[
    model_ready["START_TIME_DT"] <= train_end
].copy()

val_df = model_ready[
    (model_ready["START_TIME_DT"] > train_end)
    & (model_ready["START_TIME_DT"] <= val_end)
].copy()

test_df = model_ready[
    model_ready["START_TIME_DT"] > val_end
].copy()

print("Train:", train_df.shape)
print("Validation:", val_df.shape)
print("Test:", test_df.shape)
```

Scale the Features


```python
scaler = StandardScaler()
scaler.fit(train_df[FEATURES])

train_df.loc[:, FEATURES] = scaler.transform(
    train_df[FEATURES]
)

val_df.loc[:, FEATURES] = scaler.transform(
    val_df[FEATURES]
)

test_df.loc[:, FEATURES] = scaler.transform(
    test_df[FEATURES]
)
```

Now we build LSTM sequences

We need 3 consecutive rows->
target at next hour



```python
def create_sequences(data, features, target, lookback=3):

    X = []
    y = []

    metadata = []

    data = data.sort_values(
        ["BLOCK_ID", "START_TIME_DT"]
    )

    for block_id, block in data.groupby("BLOCK_ID"):

        block = block.sort_values("START_TIME_DT").reset_index(drop=True)

        feature_values = block[features].values
        target_values = block[target].values
        timestamps = block["START_TIME_DT"].values

        for i in range(lookback - 1, len(block)):

            # History window
            start = i - lookback + 1
            end = i + 1

            # Check that the 3 rows are exactly hourly
            history_times = pd.to_datetime(
                timestamps[start:end]
            )

            time_diffs = (
                history_times[1:] - history_times[:-1]
            )

            if not all(
                diff == pd.Timedelta(hours=1)
                for diff in time_diffs
            ):
                continue

            # Target must be next hour
            if i + 1 >= len(block):
                continue

            target_time = pd.Timestamp(
                timestamps[i + 1]
            )

            current_time = pd.Timestamp(
                timestamps[i]
            )

            if target_time - current_time != pd.Timedelta(hours=1):
                continue

            X.append(
                feature_values[start:end]
            )

            y.append(
                target_values[i + 1]
            )

            metadata.append({
                "BLOCK_ID": block_id,
                "CURRENT_TIME": current_time,
                "TARGET_TIME": target_time,
                "CURRENT_AVAILABILITY": block.loc[i, "GMP_AVAILABILITY"]
            })

    return (
        np.asarray(X, dtype=np.float32),
        np.asarray(y, dtype=np.float32),
        pd.DataFrame(metadata)
    )
```

We build the 3 dataSets

```python
X_train, y_train, meta_train = create_sequences(
    train_df,
    FEATURES,
    TARGET,
    LOOKBACK
)

X_val, y_val, meta_val = create_sequences(
    val_df,
    FEATURES,
    TARGET,
    LOOKBACK
)

X_test, y_test, meta_test = create_sequences(
    test_df,
    FEATURES,
    TARGET,
    LOOKBACK
)
```

Inspecting the shape of the 3 dataSets

```python
print("X_train:", X_train.shape)
print("y_train:", y_train.shape)

print("X_val:", X_val.shape)
print("y_val:", y_val.shape)

print("X_test:", X_test.shape)
print("y_test:", y_test.shape)
```

Now we build the LSTM. We choose the sigmoid function as the activation function since we are dealing with availability so we want an output b/w [0,1].

```python
model = keras.Sequential([
    layers.Input(
        shape=(LOOKBACK, len(FEATURES))
    ),

    layers.LSTM(
        64,
        return_sequences=False
    ),

    layers.Dropout(0.2),

    layers.Dense(
        32,
        activation="relu"
    ),

    layers.Dense(
        1,
        activation="sigmoid"
    )
])
```

Compile the model

```python
model.compile(
    optimizer=keras.optimizers.Adam(
        learning_rate=0.001
    ),

    loss="mse",

    metrics=[
        keras.metrics.MeanAbsoluteError(name="mae")
    ]
)
```

Generate summary of the Model

```python
model.summary()
```

We also add early stopping to prevent overfitting and save computation cost

```python
early_stopping = keras.callbacks.EarlyStopping(
    monitor="val_loss",
    patience= 15,
    restore_best_weights=True
)
```

Train the model

```python
history = model.fit(
    X_train,
    y_train,

    validation_data=(
        X_val,
        y_val
    ),

    epochs=100,
    batch_size=64,

    callbacks=[
        early_stopping
    ],

    verbose=1
)
```

Training set Plot

```python
plt.figure(figsize=(10, 5))

plt.plot(
    history.history["loss"],
    label="Training loss"
)

plt.plot(
    history.history["val_loss"],
    label="Validation loss"
)

plt.xlabel("Epoch")
plt.ylabel("MSE")
plt.title("LSTM Training")
plt.legend()
plt.show()
```

Evaluate on test Set

```python
test_loss, test_mae = model.evaluate(
    X_test,
    y_test,
    verbose=0
)

print("Test MSE:", test_loss)
print("Test MAE:", test_mae)
```

```python
y_pred = model.predict(
    X_test,
    verbose=0
).flatten()

mae = mean_absolute_error(
    y_test,
    y_pred
)

rmse = np.sqrt(
    mean_squared_error(
        y_test,
        y_pred
    )
)

r2 = r2_score(
    y_test,
    y_pred
)

print("MAE :", mae)
print("RMSE:", rmse)
print("R²  :", r2)
```

Note that MAE tells us how wrong on average the LSTM predictions generally are. It doesn't tell us about the error or uncertainity in individual predictions.

MC Dropouts- (Monte Carlo Dropouts. Basically we turn on the dropouts even when making predictions and then make the model predict target on same input data many times to see the variance/ uncertainity in predictions)

```python
def mc_dropout_predict(
    model,
    X,
    n_iterations=100
):

    predictions = []

    for _ in range(n_iterations):

        pred = model(
            X,
            training=True
        ).numpy().flatten()

        predictions.append(pred)

    predictions = np.asarray(
        predictions
    )

    mean_prediction = predictions.mean(
        axis=0
    )

    uncertainty = predictions.std(
        axis=0
    )

    return (
        mean_prediction,
        uncertainty,
        predictions
    )
```

```python
mc_mean, mc_std, mc_samples = mc_dropout_predict(
    model,
    X_test,
    n_iterations=100
)
```

Now we train our model on the entire dataset and output the features to be u=fed into Fuzzy Logic Machinery

First we make 3 hour sequences for every hour

```python
def create_all_prediction_sequences(
    data,
    features,
    scaler,
    lookback=3
):

    data = data.copy()

    data["START_TIME_DT"] = pd.to_datetime(
        data["START_TIME_DT"]
    )

    data = data.sort_values(
        ["BLOCK_ID", "START_TIME_DT"]
    )

    X = []
    metadata = []

    for block_id, block in data.groupby("BLOCK_ID"):

        block = block.sort_values(
            "START_TIME_DT"
        ).reset_index(drop=True)

        for i in range(lookback - 1, len(block) - 1):

# Last known hours used as input
            history = block.iloc[
                i - lookback + 1 : i + 1
            ]

# The row we are predicting
            target_row = block.iloc[i + 1]

            history_times = history[
                "START_TIME_DT"
            ].tolist()

            current_time = block.loc[
                i, "START_TIME_DT"
            ]

            target_time = target_row[
                "START_TIME_DT"
            ]

# Check that history is continuous hourly
            history_valid = all(
                history_times[j]
                - history_times[j - 1]
                == pd.Timedelta(hours=1)
                for j in range(1, len(history_times))
            )

# Check target is exactly one hour after history
            target_valid = (
                target_time - current_time
                == pd.Timedelta(hours=1)
            )

            if not history_valid or not target_valid:
                continue

# LSTM INPUT

            X_raw = history[
                features
            ]

            X_scaled = scaler.transform(
                X_raw
            )

            X.append(X_scaled)

# METADATA


            metadata.append({
                "BLOCK_ID": block_id,
                "TARGET_TIME": target_time,
                "LAST_HISTORY_TIME": current_time,

# RATE corresponding to the prediction time
                "RATE": target_row["RATE"]
            })

    X = np.asarray(
        X,
        dtype=np.float32
    )

    metadata = pd.DataFrame(
        metadata
    )

    return X, metadata
```

```python
X_all, meta_all = create_all_prediction_sequences(
    data=model_ready,
    features=FEATURES,
    scaler=scaler,
    lookback=LOOKBACK
)
```

Now we check the result

```python
print("Sequence shape:", X_all.shape)

print(
    "Number of prediction instances:",
    len(X_all)
)

print(
    "Number of blocks:",
    meta_all["BLOCK_ID"].nunique()
)

print(
    "Number of prediction times:",
    meta_all["TARGET_TIME"].nunique()
)
```

Run LSTM on all the sequences

```python
N_MC = 100 #MC Dropout rate

mc_predictions = []

for i in range(N_MC):

    predictions = model(
        X_all,
        training=True
    ).numpy().flatten()

    mc_predictions.append(
        predictions
    )

mc_predictions = np.asarray(
    mc_predictions
)

print(
    "MC prediction shape:",
    mc_predictions.shape
)
```

We calculate predicted unavailability by taking mean of 100 predictions and uncertainity byfinding the standard deviation

```python
predicted_availability = (
    mc_predictions.mean(axis=0)
)
prediction_uncertainty = (
    mc_predictions.std(axis=0)
)
```

Create the final fuzzy set by combining metaData with predictions

```python
fuzzy_dataset = meta_all.copy()

fuzzy_dataset[
    "PREDICTED_GMP_AVAILABILITY"
] = predicted_availability

fuzzy_dataset[
    "PREDICTION_UNCERTAINTY"
] = prediction_uncertainty
fuzzy_dataset = fuzzy_dataset.sort_values(
    [
        "TARGET_TIME",
        "BLOCK_ID"
    ]
).reset_index(drop=True)
```
